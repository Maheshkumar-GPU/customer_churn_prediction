LOADING THE CLEANED DATA SET

```python
import numpy as np
import pandas as pd
df = pd.read_csv('cleaned_data.csv')
```

SEPERATING THE FEATURES AND TARGET VALUE

```python
x = df.drop("Churn", axis=1)
y = df["Churn"]
y.value_counts()
```

TRAIN TEST AND SPLIT

```python
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42,stratify=y)
```

BASELINE MODEL

1) LOGISTIC REGRESSION

```python
from sklearn.linear_model import LogisticRegression
lr = LogisticRegression(max_iter=1000)
lr.fit(x_train, y_train)
y_prd = lr.predict(x_test)
```

ENSEMBLE MODEL

RANDOM FOREST

```python
from sklearn.ensemble import RandomForestClassifier
rf = RandomForestClassifier(n_estimators=100,random_state=42)
rf.fit(x_train, y_train)
rf_prd = rf.predict(x_test)
```

EVALUATION

```python
from sklearn.metrics import accuracy_score,classification_report
```

LOGISTIC REGRESSION

```python
print("logistic regression accuracy : ", accuracy_score(y_test, y_prd))
```

RANDOM FOREST

```python
print("random forest accuracy : ", accuracy_score(y_test, rf_prd))
```

CLASSIFICATION REPORT

```python
print("classification report for logistic: ", classification_report(y_test, y_prd))
print("classification report for random forest: ", classification_report(y_test, rf_prd))
```

CONFUSSION MATRIX

```python
from sklearn.metrics import confusion_matrix
yp = rf.predict(x_test)
final_prd = confusion_matrix(y_test, yp)
print(final_prd)
```

FEATURE IMPORTANCE

```python
imp = pd.Series(rf.feature_importances_,index = x.columns).sort_values(ascending=False)
imp.head(10)
```

SAVING THE MODEL

```python
import joblib
joblib.dump(rf,'churn_model.pkl')
```

SAVING THE COLUMN NAMES

```python
joblib.dump(x.columns,'feature_columns.pkl')
```
