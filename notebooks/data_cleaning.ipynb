import and load data

```python
import numpy as np
import pandas as pd
df = pd.read_csv("C:/Users/Asus/OneDrive/Documents/ml project datasets/customer churn/telecom_customer_churn.csv")
df.columns
```

target column check

```python
df["Customer Status"].value_counts()
```

to remove joined costomers

```python
df = df[df["Customer Status"] != "Joined"]
df["Customer Status"].value_counts()
```

to create target columns

```python
df["Churn"] = df["Customer Status"].map({"Stayed":0,"Churned ":1})
df["Churn"].isnull().sum()
```

drop leaking columns

```python
df.drop(columns=["Customer Status","Churn Category","Churn Reason"],inplace = True)
df.columns
```

drop ID and location columns

```python
df.drop(columns=["Customer ID",'City','Zip Code','Latitude','Longitude'],inplace = True)
df.columns
```

handeling missing values

```python
df.isnull().sum()
```

```python
net_cols = [
    'Internet Type', 'Online Security', 'Online Backup',
    'Device Protection Plan', 'Premium Tech Support',
    'Streaming TV', 'Streaming Movies', 'Streaming Music',
    'Unlimited Data', 'Avg Monthly GB Download'
]

for col in net_cols:
    df[col] = df[col].fillna('No internet service')
```

```python
 phone_cols = [
    'Avg Monthly Long Distance Charges',
    'Multiple Lines'
]

for col in phone_cols:
    df[col] = df[col].fillna('No phone service')
```

```python
df["Offer"] = df["Offer"].fillna("no offers")
```

```python
df = df.dropna(subset=['Churn'])
df.isnull().sum()
```

identifying the categorical coloumns

```python
cat_col = df.select_dtypes(include="object").columns
print(cat_col)
```

one hot encoding

```python
df = pd.get_dummies(df,columns= cat_col,drop_first=True)
```

```python
df.select_dtypes(include="object").columns

```

```python
df = df.astype(int)
```

checking the data types

```python
df.dtypes.value_counts()
```

saving the cleaned dataset

```python
df.to_csv("cleaned_data.csv",index=False)
```
